Supervised leaning example: Simple linear regression

```python
rng=np.random.RandomState(42)
x=10*rng.rand(50)
x
y=2*x-1+rng.randn(50)
y
plt.scatter(x, y);
```

```python
#choosing class of model
from sklearn.linear_model import LinearRegression 
#choose model hyperparameters
model = LinearRegression (fit_intercept=True)
model
```

```python
#Arranging data into a features matrix and target vector
x = x[:, np.newaxis]
x
```

```python
#fit the model to your data
model.fit(x,y)
#These two parameters represent the slope and intercept of the simple linear fit to the data
print (model.coef_)
print (model.intercept_)
```

```python
#predict labels for unknown data 
xfit = np.linspace(-1,11)
Xfit = xfit[:, np.newaxis]
yfit = model.predict(Xfit)
yfit
```

```python
#visualising result using plot, plotting the raw data and the model fit
plt.scatter(x, y)
plt.plot(xfit,yfit)
```

```python
import seaborn as sns
iris = sns.load_dataset('iris')
iris.head(20) #brings out the first species
# iris.tail()  #brings out the last species
```

```python
print (iris.shape)
#iris.species.unique()
iris['species'].unique()
#iris.species
```

```python
%matplotlib inline
# sns.set()
sns.pairplot(iris, hue = 'species', height=1.5);
plt.show()
```

```python
X_iris =iris.drop('species', axis=1)
X_iris.shape
```

```python
y_iris=iris['species']
y_iris.shape
```

```python
#spliting data to training and testing set
#This could only be done by hand, but better to use the train_test_split func:

from sklearn.model_selection import train_test_split
Xtrain, Xtest, ytrain, ytest = train_test_split(X_iris, y_iris, random_state=1, train_size=0.8)
```

```python
Xtrain.shape
```

```python
from sklearn.naive_bayes import GaussianNB #1. choose model class
model = GaussianNB() #2. instantiate model
model.fit(Xtrain, ytrain) #fit model to data
y_model = model.predict(Xtest) #4. predict on new data
```

```python
print ("Predicted data: \n", y_model)
print ("Origial data: \n", ytest.values)
```

```python
#convert string to int, because ML can only read int
ir['species']=ir['species'].astype('category').cat.codes
ir['species']
```

```python
x = ir.drop('species', axis =1)
x
```

```python
y = ir['species']
```

```python
#splitting to train and test (20% for test)
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(x,y, test_size = 0.2, random_state = 24)
```

```python
from sklearn import svm
from sklearn.linear_model import LogisticRegression
```

```python
lg=LogisticRegression()
model_log=lg.fit(X_train,y_train)
```

```python
print('predicted: ', model_log.predict(X_test))
print ("")
print('original: ', np.array(y_test))
```

```python
#for accuracy
model_log.score(X_test, y_test)
```
